## Part 5: Construct the Maximum Return Portfolio

Compute the maximum return portfolio based on your previously estimated risks and returns.

```python
# YOUR CHANGES HERE

%pip install cvxpy
import cvxpy as cp

# Compute maximum return
n = len(expected_returns)
x = cp.Variable(n)

objective = cp.Maximize(expected_returns.to_numpy().reshape(1, -1) @ x)
# objective

prob = cp.Problem(objective, [0 <= x, cp.sum(x) == 1])

estimated_return_P = prob.solve()
# estimated_return_P

x.value.round(2)

# Construct maximum return portfolio
maximum_return_portfolio = pd.DataFrame({
    "Asset" : expected_returns.index,
    "Allocation" : x.value.round(2),
})

maximum_return_portfolio.set_index("Asset", inplace=True)
maximum_return_portfolio
```

Compute the minimum risk portfolio based on your previously estimated risks.

```python
# YOUR CHANGES HERE

# Compute minimum return
n = len(expected_returns)
x_minimum_risk = cp.Variable(n)

objective_minimum_risk = cp.Minimize(x_minimum_risk.T @ covariance_matrix.to_numpy() @ x_minimum_risk)
# objective_minimum_risk

prob_minimum_risk = cp.Problem(objective_minimum_risk, [0 <= x_minimum_risk, cp.sum(x_minimum_risk) == 1])

covariance_minimum_risk = prob_minimum_risk.solve()
# covariance_minimum_risk

x_minimum_risk.value.round(2)

# Construct minimum risk portfolio
minimum_risk_portfolio = pd.DataFrame({
    "Asset" : expected_returns.index,
    "Allocation" : x_minimum_risk.value.round(2),
})

# minimum_risk_portfolio.set_index("Asset", inplace=True)
minimum_risk_portfolio
```

## Part 7: Build Efficient Frontier Portfolios

Compute 101 portfolios along the mean-variance efficient frontier with evenly spaced estimated returns.
The first portfolio should be the minimum risk portfolio from part 4, and the last portfolio should be the maximum return portfolio from part 3.
The estimated return of each portfolio should be higher than the previous by one percent of the difference between the first and last portfolios.
That is, the estimated return of the portfolios should be similar to `np.linspace(min_risk_return, max_return, 101)`.


```python
# YOUR CHANGES HERE

# Compute Efficient Frontier Portfolios
estimated_return_minimum_risk = x_minimum_risk.value @ expected_returns
# estimated_return_minimum_risk
estimated_return_maximum_return = expected_returns.max()
# estimated_return_maximum_return


# 101 optimized portfolios
efficient_variances = []
efficient_returns = []
efficient_portfolios = []

for r in np.linspace(estimated_return_minimum_risk, estimated_return_maximum_return, 101):
    x_r = cp.Variable(n)

    prob_r = cp.Problem(cp.Minimize(x_r.T @ covariance_matrix.to_numpy() @ x_r), 
                        [0 <= x_r, cp.sum(x_r) == 1, expected_returns.to_numpy().reshape(1, -1) @ x_r == r])

    efficient_variances.append(prob_r.solve())
    efficient_returns.append(r)
    efficient_portfolios.append(x_r.value)

efficient_portfolios = np.asarray(efficient_portfolios).round(2)
# efficient_portfolios


# Construct DataFrame
efficient_portfolios_df = pd.DataFrame({
    "Index" : range(1, len(efficient_portfolios) + 1),
    "Return": efficient_returns,
    "Risk": np.sqrt(efficient_variances),
    "AAPL" : efficient_portfolios[:, 0],
     "SPY" : efficient_portfolios[:, 1],
     "TSLA" : efficient_portfolios[:, 2],
     "EURUSD" : efficient_portfolios[:, 3],
     "META" : efficient_portfolios[:, 4],
     "SPOT" : efficient_portfolios[:, 5]
})
efficient_portfolios_df

# VISUALS:
#
# 1. Portfolio Performance
# plt.plot(np.sqrt(covariance_matrix), expected_returns, label="Assets", linestyle="none", marker="o")
# plt.plot(np.sqrt(efficient_variances), efficient_returns, label="Efficient Frontier", linestyle="--")
#
# plt.legend()
# plt.xlabel("Risk (Standard Deviation)")
# plt.ylabel("Expected Return")
# plt.title("Mean-Variance Efficient Frontier and Individual Assets")
# plt.xlim(0)
# plt.ylim(0)
# plt.show()
#
# 2. Portfolio Composition Map
# plt.figure(figsize=(10, 5))
#
# bottom = np.zeros(efficient_portfolios.shape[0])
# for (i, a) in enumerate(historical_prices.columns):
#     plt.bar(x=efficient_returns,
#             height=efficient_portfolios[:, i],
#             bottom=bottom, 
#             label=a,
#             width=(efficient_returns[1]-efficient_returns[6]))
#     bottom += efficient_portfolios[:, i]
#
# plt.legend()
# plt.xlabel("Portfolio Expected Return")
# plt.xlim(estimated_return_minimum_risk, estimated_return_maximum_return)
# plt.ylabel("Portfolio Allocation")
# plt.title("Portfolio Composition Map of the Mean-Variance Efficient Frontier")
# plt.show()
```

## Part 8: Check Maximum Return Portfolio Stability

Check the stability of the maximum return portfolio by resampling the estimated risk/return model.

Repeat 1000 times -
1. Use `np.random.multivariate_normal` to generate 23 return samples using your previously estimated risks and returns.
2. Estimate the return of each asset using that resampled return history.
3. Check which asset had the highest return in those resampled estimates.

This procedure is a reduced and simplified version of the Michaud resampled efficient frontier procedure that takes uncertainty in the risk model into account.

```python
# YOUR CHANGES HERE

# Check stability of maximum return portfolio by resampling the model
n_repeats = 1000
n_samples = 23

# Generate 23 return samples
assets = expected_returns.index
mean_vec = expected_returns.to_numpy()
cov_mat = covariance_matrix.to_numpy()

best_asset_counts = []

for n in range(n_repeats):
    simulated_returns = np.random.multivariate_normal(mean_vec, cov_mat, size=n_samples)

    # estimated return of each asset
    resampled_estimated_returns = simulated_returns.mean(axis=0)

    # asset with highest return in resampled estimates
    best_asset = assets[np.argmax(resampled_estimated_returns)]
    best_asset_counts.append(best_asset)

best_asset_counts = pd.DataFrame(best_asset_counts)

# Construct DataFrame
best_asset_counts_df = pd.DataFrame({
    'Asset': best_asset_counts[0].value_counts().index,
    'Probability': best_asset_counts[0].value_counts(normalize=True).values
})
best_asset_counts_df.set_index('Asset', inplace=True)
best_asset_counts_df
```

The header row should have columns "asset" and "probability".
There should be a data row for each asset and its sample probability of having the highest return based on those 1000 resampled estimates.
